Deep dive into PyTorch model logits and thresholds.

```python
from pathlib import Path

import numpy as np
import torch

from panoseti_analysis.io.models import load_classifier
from panoseti_analysis.paths import MODELS
```

## Model Inspection

```python
model, bundle = load_classifier(MODELS / "cloud_detector_v1.pt")
model.eval()

# Create a dummy batch of features (N, 2, 32, 32)
dummy_input = torch.randn(4, 2, 32, 32)
with torch.no_grad():
    logits = model(dummy_input)
    probs = torch.nn.functional.softmax(logits, dim=1)

print("Logits:\n", logits)
print("Probs:\n", probs)
```
