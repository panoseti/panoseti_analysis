## Alternative: pure-Python dev driver

For notebook R&D and smoke-testing without Nextflow, `run_pipeline()` mirrors the same
ingest + ML sequence as a flat Python function. It writes the same `L0/`, `L1/`, `L2/`
layout and returns typed `StoreLineage` records.

| | `nextflow run .` | `run_pipeline()` |
|---|---|---|
| Resume / SLURM / containers | ✓ | ✗ |
| No Nextflow install needed | ✗ | ✓ |
| Parallelism | ✓ | ✗ (sequential) |
| Typed Python return values | ✗ | ✓ |

See **Tutorial 5** for a full walkthrough.

```python
from panoseti_analysis.adapters.recipe_driver import run_pipeline

OBS_DIR = REPO_ROOT / "tests" / "data" / "obs_TEST.pffd"
OUT_DIR = REPO_ROOT / "notebooks" / "tutorials" / "results_tutorial_py"

outputs = run_pipeline(OBS_DIR, OUT_DIR)

print(f"L0 stores : {len(outputs.l0_stores)}")
print(f"L1 stores : {len(outputs.l1_stores)}")
for rec in outputs.l1_stores:
    print(f"  {rec.store}  kind={rec.kind}  frames={rec.n_frames}")
```
