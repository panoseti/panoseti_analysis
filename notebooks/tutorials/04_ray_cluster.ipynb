# Tutorial 4: Connecting to Ray

How to connect to a Ray cluster to run ML jobs.

```
import ray
import xarray as xr
from panoseti_analysis.io.stores import open_store
```

## Initialize Ray

```
ray.init(ignore_reinit_error=True)
print("Ray connected!", ray.cluster_resources())
```

## Remote Execution
We can define remote functions and execute them over our stores.

```
@ray.remote
def get_mean_l1(store_path):
    ds = xr.open_zarr(store_path, consolidated=False)
    return float(ds["median_subtracted"].mean().values)

l1_path = "../../results_tutorial/L1/obs_TEST.dp_img16.module_1.L1.zarr"
future = get_mean_l1.remote(l1_path)
mean_val = ray.get(future)
print("Mean value:", mean_val)
```
