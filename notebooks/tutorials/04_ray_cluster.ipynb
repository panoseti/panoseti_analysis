## Remote Execution
We can define remote functions and execute them over our stores.

```python
!pwd
```

```python
@ray.remote
def get_mean_l1(ds_ref):
    
    return float(ds_ref["median_subtracted"].mean().values)

l1_path = "results_tutorial/L1/obs_TEST.dp_img16.module_1.L1.zarr"
ds = xr.open_zarr(l1_path, consolidated=False)
ds_ref = ray.put(ds)
future = get_mean_l1.remote(ds_ref)
mean_val = ray.get(future)
print("Mean value:", mean_val)
```
